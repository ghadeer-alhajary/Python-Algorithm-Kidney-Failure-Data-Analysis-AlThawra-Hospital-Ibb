# # إضافة عمود جديد باسم "nazeh"

```python

# تحميل البيانات من ملف CSV
df = pd.read_csv('KidneyPatientsGHadeer.csv', sep=';')
```

```python

# إضافة عمود جديد باسم "nazeh"
# يحتوي على "YAS" إذا كانت الخلية تقع ضمن صفوف العمود "Sickyear" وتحتوي على القيمة "2018-n" أو "2019-n"، و"NO" في جميع الحالات الأخرى
def check_sick_year(row):
    sick_year = row['Sickyear']
    if isinstance(sick_year, str) and (sick_year.startswith('2018-') or sick_year.startswith('2019-')):
        return 'YAS'
    else:
        return 'NO'

df['nazeh'] = df.apply(check_sick_year, axis=1)
```

```python

# حفظ البيانات المحدثة في ملف CSV
df.to_csv('KidneyPatientsGHadeer.csv', index=False)
```

```python

# إضافة عمود جديد باسم "Number"
# يحتوي على رقم التسلسلي لكل صف في الجدول
def add_code_number(row):
    return row.name + 1

df['Number'] = df.apply(add_code_number, axis=1)
```

# ولأني أعمل على بيانات عربية فأسماء المدن والمناطق مؤنث غير عاقل لهذا يجب انتهائها بتاء مربوطة وليس بهاء ولهذا سأحل مشكلة التاء المربوطة

```python
import pandas as pd
import re
```

```python

# Define a function to replace "ه" with "ة" at the end of each word
def replace_h_to_taa(word):
    pattern = re.compile(r'\b(\w+)ه(\W|$)')
    return pattern.sub(r'\1ة\2', word)
```

```python

# Apply the function to the "Directorate" column
df['Directorate'] = df['Directorate'].apply(lambda x: ' '.join(replace_h_to_taa(word) if word.endswith('ه')
                                                               else word for word in x.split()))
```

```python

# Apply the function to the "Governorate" column
df['Governorate'] = df['Governorate'].apply(lambda x: ' '.join(replace_h_to_taa(word) if word.endswith('ه')
                                                         else word for word in x.split()))
```
